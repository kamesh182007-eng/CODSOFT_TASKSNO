```python
import pandas as pd

data = {
    "Name": ["Arun", "Divya", "Karthik", "Meena", "Ravi", "Sneha", "Vijay", "Priya", "Suresh", "Lakshmi"],
    "Gender": ["M", "F", "M", "F", "M", "F", "M", "F", "M", "F"],
    "Department": ["CSE", "ECE", "CSE", "IT", "ECE", "CSE", "IT", "IT", "ECE", "CSE"],
    "Attendance": [92, 85, 70, 95, 60, 88, 75, 98, 65, 90],
    "Maths": [78, 82, 55, 91, 40, 85, 62, 95, 48, 88],
    "Python": [85, 79, 60, 94, 45, 90, 68, 97, 50, 92],
    "Physics": [74, 88, 52, 89, 42, 80, 65, 93, 55, 84],
}

df = pd.DataFrame(data)
df["Total"] = df[["Maths", "Python", "Physics"]].sum(axis=1)
df["Average"] = (df["Total"] / 3).round(1)
df
```

```python
df.describe()
```

Mean Maths is 72.4 and mean attendance is 81.8. A few low scores pull the average below the median.

```python
df.groupby("Department")[["Attendance", "Maths", "Python", "Physics", "Average"]].mean().round(1)
```

IT has the highest average (83.8) and ECE the lowest (58.8). Departments with higher attendance also score higher.

```python
df[["Attendance", "Maths", "Python", "Physics", "Average"]].corr().round(2)
```

Attendance has a very strong positive correlation with average marks (0.98). Students who attend more tend to score higher.

```python
df[(df["Attendance"] < 70) | (df["Average"] < 60)][["Name", "Department", "Attendance", "Average"]]
```

Three students (Karthik, Ravi, Suresh) have low attendance and low marks. Two are from ECE. Ravi is the biggest outlier with an average of 42.3.

## Summary of Findings
1. The class has 10 students and no missing values.
2. Mean Maths is 72.4, mean attendance is 81.8, and mean average marks are 73.5.
3. IT has the best average (83.8); ECE has the lowest (58.8).
4. Attendance and average marks have a very strong correlation (0.98).
5. Three students are at risk: Karthik, Ravi and Suresh.
